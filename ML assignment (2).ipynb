```python
import numpy as np

# Given dataset
x_1 = np.array([1, 2, 3])
x_2 = np.array([3, 2, 3])
y = np.array([2, 3, 4])

# Initialize parameters
d_theta_0 = 0
d_theta_1 = 1
d_theta_2 = 2

alpha = 0.1  # Learning rate
epochs = 10
m = len(x_1)  # Number of data points

# Gradient Descent
for i in range(epochs):
    # Compute predictions
    h = d_theta_0 + (d_theta_1 * x_1) + (d_theta_2 * x_2)

    # Compute errors
    error = h - y

    # Compute partial derivatives
    d_theta_0 = (1/m) * np.sum(error)
    d_theta_1 = (1/m) * np.sum(error * x_1)
    d_theta_2 = (1/m) * np.sum(error * x_2)

    # Update parameters
    d_theta_0 -= alpha * d_theta_0
    d_theta_0 -= alpha * d_theta_1
    d_theta_0 -= alpha * d_theta_2

    # Print updated values
    print(f"Iteration {i+1}: theta_0 = {d_theta_0:.4f}, theta_1 = {d_theta_1:.4f}, theta_1 = {d_theta_2:.4f}")

# Final parameters
print("Final values:")
print(f"theta_0 = {d_theta_0:.4f}")
print(f"theta_1 = {d_theta_1:.4f}")
print(f"theta_1 = {d_theta_2:.4f}")
```
