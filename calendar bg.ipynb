```python
import calendar
import matplotlib.pyplot as plt
from PIL import Image
import datetime
```

```python
bg = Image.open("nyc.jpg")
```

```python
plt.figure(figsize=(10, 7))
plt.imshow(bg)
plt.axis("off")
```

```python
plt.imshow(bg)
plt.axis("off")

# Proper overlay using axis coordinates
overlay = plt.Rectangle(
    (0, 0),
    1,
    1,
    transform=plt.gca().transAxes,
    color="black",
    alpha=0.45
)
plt.gca().add_patch(overlay)
```

```python
plt.imshow(bg)
plt.axis("off")

plt.gca().add_patch(
    plt.Rectangle(
        (0, 0),
        1,
        1,
        transform=plt.gca().transAxes,
        color="black",
        alpha=0.45
    )
)
```

```python
fig, ax = plt.subplots(figsize=(10, 7))

ax.imshow(bg)
ax.axis("off")

month_name = datetime.date(year, month, 12).strftime("%B %Y")

ax.text(
    0.5,
    0.88,
    month_name,
    fontsize=28,
    color="white",
    ha="center",
    va="center",
    transform=ax.transAxes,
    fontweight="bold"
)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 7))

ax.imshow(bg)
ax.axis("off")

# Calendar text
ax.text(
    0.5,
    0.5,
    cal,
    fontsize=20,
    color="yellow",
    ha="center",
    va="center",
    transform=ax.transAxes,
    family="monospace"
)

plt.show()
```
